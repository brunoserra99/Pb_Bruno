```python
import random
import csv

lista = [random.randint(1, 100000)for _ in range(250)]
print(*lista, sep="\n")

#salvando em arquivo .csv
with open("/content/lista.csv", mode="w", newline="") as arquivo:
    escritor = csv.writer(arquivo)

    #cabecalho
    escritor.writerow(["numeros"])

    #lista
    for numero in lista:
        escritor.writerow([numero])

print("Arquivo 'lista.csv' foi criada com sucesso")
print(lista)
```

```python
# Passo 2: Importar bibliotecas necessárias
import random
import time
import os
import names

# time
start_time = time.time()

# Passo 3: Definir os parâmetros para geração do dataset
random.seed(40)
qtd_nomes_unicos = 3000
qtd_nomes_aleatorios = 10000000

# Passo 4: Gerar os nomes únicos
aux = []
for i in range(qtd_nomes_unicos):
    aux.append(names.get_full_name())
print(f"Gerando {qtd_nomes_aleatorios} nomes aleatórios...")

dados = []
for i in range(qtd_nomes_aleatorios):
    dados.append(random.choice(aux))

# Passo 5: Salvar os nomes em um arquivo de texto
nome_arquivo = "/content/nomes_aleatorios.txt"
with open(nome_arquivo, mode="w", newline="") as arquivo:
    for nome in dados:
        arquivo.write(f"{nome}\n")
print(f"Arquivo '{nome_arquivo}' foi criado com sucesso!")

# Passo 6: Abrir e verificar o conteúdo do arquivo
with open(nome_arquivo, mode="r") as arquivo:
    conteudo = arquivo.readlines()[:20]
    print("\nExemplo de conteúdo do arquivo:")
    for linha in conteudo:
        print(linha.strip())

# time
end_time = time.time()
# Exibir o tempo de execução
print(f"Dataset gerado com sucesso! Tempo de execução: {end_time - start_time:.2f} segundos")
```

Criando e iniciando a SparkSession

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lit, when, rand

spark = SparkSession\
    .builder\
    .master("local[*]")\
    .appName("Exercicio Intro")\
    .getOrCreate()

print("Iniciado SparkSession")
```

Inicializando o arquivo

```python
df_nomes = spark.read.csv("/content/nomes_aleatorios.txt", header=False, inferSchema=True)
df_nomes.show(5)
```

Alterando o nome da coluna _c0 para "nome"

```python
#df_nomes.printSchema()
#df_nomes = df_nomes.withColumnRenamed("_c0", "nome")
#df_nomes.show(10)
#df_nomes['nome']
df_nomes.nome
```

Adicionando a coluna escolaridade e a populando

```python
df_nomes = df_nomes.withColumn("escolaridade", when(rand() < 0.33, "Ensino Médio")
    .when(rand() < 0.66, "Ensino Superior")
    .otherwise("Pós-graduação"))
df_nomes.show(5)
```

Adicioonado a coluna paises e a populando

```python
from pyspark.sql.functions import expr

paises_expr = """CASE
    WHEN rand() < 1/13 THEN 'Brasil'
    WHEN rand() < 2/13 THEN 'Argentina'
    WHEN rand() < 3/13 THEN 'Chile'
    WHEN rand() < 4/13 THEN 'Uruguai'
    WHEN rand() < 5/13 THEN 'Paraguai'
    WHEN rand() < 6/13 THEN 'Bolívia'
    WHEN rand() < 7/13 THEN 'Peru'
    WHEN rand() < 8/13 THEN 'Equador'
    WHEN rand() < 9/13 THEN 'Colômbia'
    WHEN rand() < 10/13 THEN 'Venezuela'
    WHEN rand() < 11/13 THEN 'Suriname'
    WHEN rand() < 12/13 THEN 'Guiana'
    ELSE 'Guiana Francesa'
END"""

df_nomes = df_nomes.withColumn("pais", expr(paises_expr))
df_nomes.show(15)
```

Adicionando a coluna ano e a populando

```python
df_nomes = df_nomes.withColumn("ano_nasc", (lit(1945) + (rand() * (2010 - 1945)).cast("int")))
df_nomes.show(15)
```

Filtro de nascidos neste seculo

```python
df_select = df_nomes.filter(df_nomes.ano_nasc >= 2000)
df_select.show(10)
```

Filtro de nascidos neste seculo (sparkSQL)

```python
df_nomes.createOrReplaceTempView("pessoas")
df_sql = spark.sql("SELECT * FROM pessoas WHERE ano_nasc >= 2000")
df_sql.show(10)
```

Filtro geraçao millennials

```python
qtd_millennials = df_nomes.filter((df_nomes.ano_nasc >= 1980) & (df_nomes.ano_nasc <= 1994)).count()
print("Número de millennials:", qtd_millennials)
```

Filtro geraçao millennials (sparkSQL)

```python
df_millennials = spark.sql("SELECT COUNT(*) as qtd FROM pessoas WHERE ano_nasc BETWEEN 1980 AND 1994")
df_millennials.show()
```

Filtro quantidade de cada geracao por, pais/geraçao/quantidade

```python
df_geracao = spark.sql("""
    SELECT pais,
           CASE
               WHEN ano_nasc BETWEEN 1944 AND 1964 THEN 'baby boomers'
               WHEN ano_nasc BETWEEN 1965 AND 1979 THEN 'geracao x'
               WHEN ano_nasc BETWEEN 1980 AND 1994 THEN 'millennials'
               WHEN ano_nasc BETWEEN 1995 AND 2015 THEN 'geracao z'
           END AS geracao,
           COUNT(*) AS quantidade
    FROM pessoas
    GROUP BY pais, geracao
    ORDER BY pais, geracao, quantidade
""")

df_geracao.show()
```
