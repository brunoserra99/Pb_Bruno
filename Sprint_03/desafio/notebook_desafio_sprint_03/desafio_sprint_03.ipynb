# 1 - Leia o arquivo googleplaystore.csv e realize a seguinte atividade sobre o dataset utilizando as bilbliotecas e matplotlib. remova as linhas duplicadas


```python
# Função para remover duplicados
import pandas as pd

# Ler o arquivo CSV
dados = pd.read_csv('googleplaystore.csv')
linhas_antes = len(dados)

# Remover duplicados
dados_sem_duplicados = dados.drop_duplicates()
linhas_apos = len(dados_sem_duplicados)

# Salvar o novo arquivo
dados_sem_duplicados.to_csv('googleplaystore1.csv', index=False)
print(f"Linhas removidas: {linhas_antes - linhas_apos}")
print("----Finalizado c sucesso----")
```

# 2 - faça um grafico do barra contendo os top 5 apps por numero de instalcao

```python

# Função para criar gráfico de barras (top 5 apps por número de instalações)

import pandas as pd
import matplotlib.pyplot as plt

# Ler o arquivo CSV
arquivo = "googleplaystore1.csv"
dados = pd.read_csv(arquivo)

dados['Installs'] = dados['Installs'].str.replace(r'[^\d]', '', regex=True)
dados['Installs'] = dados['Installs'].replace('', '0').astype(int)
top_5_apps = dados[['App', 'Installs']].sort_values(by='Installs', ascending=False).head(5)

# Plotar gráfico de barras
plt.figure(figsize=(10,6))
plt.barh(top_5_apps['App'], top_5_apps['Installs'], color='skyblue')
plt.xlabel('Número de Instalações')
plt.title('Top 5 Apps por Número de Instalações')
plt.gca().invert_yaxis()
plt.show()
```

# 3 - faca um grafico de pizza (pie chart) mostrando os categorias de apps existente no dataset de acordo com a frequencia em que elas aparecem

```python
# Função para criar gráfico de pizza (categorias de apps)
# Importando as bibliotecas necessárias
import pandas as pd
import matplotlib.pyplot as plt

# Carregando o arquivo CSV
arquivo = 'googleplaystore1.csv'
dados = pd.read_csv(arquivo)

# Contando a frequência das categorias de apps
frequencia_categorias = dados['Category'].value_counts()

# Criando o gráfico de pizza
plt.figure(figsize=(8,8))
plt.pie(frequencia_categorias, labels=frequencia_categorias.index, autopct='%1.1f%%', startangle=90)
plt.title('Distribuição das Categorias de Apps')
plt.axis('equal')  # Deixa o gráfico de pizza circular
plt.show()
```

# 4 - mostre qual app masi caro existente no dataset

```python
# Importando as bibliotecas necessárias
import pandas as pd

# Carregando o arquivo CSV
arquivo = 'googleplaystore1.csv'
dados = pd.read_csv(arquivo)

# Limpando a coluna 'Price' para remover caracteres indesejados
dados['Price'] = dados['Price'].replace('[\$,]', '', regex=True)

# Convertendo a coluna 'Price' para valores numéricos, tratando valores não numéricos como NaN
dados['Price'] = pd.to_numeric(dados['Price'], errors='coerce')

# Encontrando o app mais caro
app_mais_caro = dados.loc[dados['Price'].idxmax()]

# Mostrando apenas as colunas 'App' e 'Price' de forma limpa
print(f"App mais caro: {app_mais_caro['App']}, Preço: {app_mais_caro['Price']}")

```

#6 - mostre o top 10 apps por numero de reviews bem como o respectivo numero de reviews, ordene a lista de forma decrescente por numero de reviews

```python
# Função para top 10 apps por número de reviews
import pandas as pd

# Carregar o arquivo CSV
df = pd.read_csv('googleplaystore1.csv')

# Converter a coluna 'Reviews' para numérica (pode conter valores não numéricos)
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

# Ordenar os aplicativos por número de reviews em ordem decrescente
top_10_reviews = df[['App', 'Reviews']].sort_values(by='Reviews', ascending=False).head(10)

# Redefinir os índices para valores sequenciais
top_10_reviews = top_10_reviews.reset_index(drop=True)

# Mostrar o resultado
print(top_10_reviews)

```

#7 - crie pelo menos mais 2 calculos sobre o dataset e apresente um em formato de lista e outro em formato de valor. por exemplo "top 10 apps por numero de reviews" e "app mais caro existente no dataset"

```python
import pandas as pd

# Carregar o arquivo CSV
df = pd.read_csv('googleplaystore1.csv')

# Converter as colunas para tipos adequados
df['Reviews'] = pd.to_numeric(df['Reviews'], errors='coerce')

# Limpar a coluna 'Price' para remover símbolos e converter para float
df['Price'] = df['Price'].replace({'\$': '', ',': ''}, regex=True)

# Filtrar apenas as linhas onde 'Price' é numérico
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

# Calcular o Top 10 aplicativos por número de reviews
top_10_reviews = df[['App', 'Reviews']].sort_values(by='Reviews', ascending=False).head(10)
top_10_reviews = top_10_reviews.reset_index(drop=True)

# Salvar o Top 10 em um arquivo CSV
top_10_reviews.to_csv('top10.csv', index=False)

# Mostrar os resultados
print("Top 10 aplicativos por número de reviews:")
print(top_10_reviews)

# Calcular o total de reviews dos top 10 aplicativos
total_reviews = top_10_reviews['Reviews'].sum()

# Calcular o total de reviews de todos os aplicativos
total_reviews_all_apps = df['Reviews'].sum()

# Mostrar o total de reviews e o total de visualizações de todos os aplicativos antes da média
print(f"\nTotal de Reviews dos Top 10 aplicativos: {total_reviews}")
print(f"Total de Visualizações de Todos os Apps: {total_reviews_all_apps}")

# Calcular a média de reviews de todos os aplicativos
media_reviews = df['Reviews'].mean()

# Mostrar a média de reviews ao final
print(f"\nMédia de Reviews de todos os aplicativos: {media_reviews:.2f}")

# Identificar os 3 aplicativos mais caros
top_3_expensive_apps = df[['App', 'Price']].sort_values(by='Price', ascending=False).head(3)

# Salvar os 3 aplicativos mais caros em um arquivo CSV
top_3_expensive_apps.to_csv('top3.csv', index=False)

# Calcular a média de preço dos 3 aplicativos mais caros
media_price_top_3 = top_3_expensive_apps['Price'].mean()

# Mostrar os 3 aplicativos mais caros e a média de preço deles
print('\n-------apps mais caro, e a media dos 03------')
print("\nOs 3 aplicativos mais caros:")
print(top_3_expensive_apps)
print(f"\nMédia de preço dos 3 aplicativos mais caros: {media_price_top_3:.2f}")
```

#8 - crie pelo menos outras 2 forma graficas de exibiçao dos indicadores acima utilizando a biblioteca matplotlib escolha graficos diferentes dos explorados acima. por exemplo linhas e dispersão

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams

# Ajustar a fonte para uma fonte mais comum
rcParams['font.family'] = 'DejaVu Sans'  # Fonte genérica, ampla compatibilidade

# Carregar os dados dos arquivos CSV
top10_df = pd.read_csv('top10.csv')
top3_df = pd.read_csv('top3.csv')

# Modificar o nome dos aplicativos problemáticos (remover caracteres especiais)
top3_df['App'] = top3_df['App'].replace({
    "I'm Rich/Eu sou Rico/أنا غني/我很有錢": "I'm Rich/Eu sou Ric..."
})

# 1. Gráfico de Linha (Line Plot) para Reviews dos Top 10 Apps
plt.figure(figsize=(10,6))
plt.plot(top10_df['App'], top10_df['Reviews'], marker='o', linestyle='-', color='b', label='Reviews dos Apps')
plt.title('Evolução dos Reviews dos Top 10 Aplicativos', fontsize=14)
plt.xlabel('Aplicativos', fontsize=12)
plt.ylabel('Número de Reviews', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()

# 2. Gráfico de Dispersão (Scatter Plot) para Preço x Reviews dos 3 Aplicativos Mais Caros
plt.figure(figsize=(10,6))
plt.scatter(top3_df['App'], top3_df['Price'], color='r', s=100, label='Preço dos Apps', marker='D')
plt.title('Preço dos 3 Aplicativos Mais Caros', fontsize=14)
plt.xlabel('Aplicativos', fontsize=12)
plt.ylabel('Preço', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```
